# **Stemming in NLP**

Stemming is the process of reducing words to thier root/base form (stem) by removing or suffixes.

# **Important**
spacy does not provide a built-in stemming algorithm.
spacy primarily uses lemmatization if you specifically want
stemming while using spacy you can intergrate
a stemmer such as NLTK's porter Stemmer or SnowballStemmer.

```python
# 1: Basic Stemming
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["playing","played","plays","player"]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
# 2: Stemming a Sentence
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
text="The Students are studying different subjects."
words=text.split()

for word in words:
  print(word,"->",stemmer.stem(word))
words=["playing","played","plays","player"]

for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
# 5: Words ending with "s"
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["cats","dogs","books","cars","students"]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
# 6: Different forms of the sale word
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["connect","connected","connecting","connection","connections"]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
# 8: Stemming words related to organize
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["organize","organized","organizing","organization","organizations"]
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
# 9: stemming text using regular expressions

import re
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
text="Students are learning python and developing machine learning projects."
words=re.findall(r"\b[a-zA-Z]+\b",text.lower())
for word in words:
  print(word,"->",stemmer.stem(word))
```

```python
# 10: Create a stemmer sentence
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
text="The platers are playing football and practicing regularly."
words=text.lower().split()
stemmed_words=[]
for word in words:
  stemmed_words.append(stemmer.stem(word))
stemmed_text = " ".join(stemmed_words)
print("Original:")
print(text)

print("\nStemmed:")
print(stemmed_text)
```

```python
#Import the libraries:
import spacy
from nltk.stem import PorterStemmer
nlp=spacy.load("en_core_web_sm")
stemmer=PorterStemmer()
```

```python
#1: Stemming spaCy Tokens

text="The students are studing."
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#2: Playing, played, plays

text="I play football. He plays football. they played yesterday."
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#3: Reading-related words

text="Reading readers read readable books."
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#4: Learning-related words

text="Students learn learning learned concepts."
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#5: Connect-related words

text="The system connects connected devices."
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#6: stemming a complete paragraph

text=""" Data scientists analyze data to extract insights.
They use various techniques, including machine learning,
to build predictive models. The process involves cleaning,
transforming, and visualizing data to make informed decisions."""
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

```python
#7: Remove punctuation before stemming

text="Python is powerful,useful,and popular!"
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#8: Stemming with lowercase conversion

text="Python Pythonic Programming Programs"
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    word=token.text.lower()
    print(word,"->",stemmer.stem(word))
```

```python
#9: Create a stemmed sentence using spaCy.

text="The children are playing games and enjoying themseives."
doc=nlp(text)
stemmer_words=[]
for token in doc:
    if token.is_alpha:
        stemmer_words.append(stemmer.stem(token.text.lower()))
stemmed_sentence=" ".join(stemmer_words)
print("Original:")
print(text)

print("\nStemmer:")
print(stemmer_words)
```
