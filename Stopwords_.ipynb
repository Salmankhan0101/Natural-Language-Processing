Stopwords are common words that ofter carry little useful information for certain NLP tasks. Exaples include:

the, is, am, are, a, an, the, and, or, in, on, of, to, for

```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
# 1: Check Whether a word is a Stopword.

words=["the","is","Python","and","beautiful"]
for word in words:
  doc=nlp(word)

  for token in doc:
    print(token.text,"->",token.is_stop)
```

```python
# 2: identify Stopwords in a sentence.

text="Python is a very useful programming language."
doc=nlp(text)
for token in doc:
  print(token.text,"->",token.is_stop)
```

```python
# 3: Extract Only Stopwords.

text="This is a simple example of stopword removal."
doc=nlp(text)
stopwords=[]
for token in doc:
  if token.is_stop:
    stopwords.append(token.text)
print(stopwords)
```

```python
# 4: Extract Non-Stopword.

text="Python is an easy and powerful language."
doc=nlp(text)
words=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print(words)
```

```python
# 5: Remove Stopwords  from a Sentence.

text="The students are learning data science in the classroom."
doc=nlp(text)
result=[]
for token in doc:
  if not token.is_stop:
    result.append(token.text)
print(" ".join(result))
```

```python
# 6: Remove Stopwords and Punctuation.

text="The students are learning data science in the classroom."
doc=nlp(text)
result=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text)
print(" ".join(result))
```

```python
# 7: Convert Text to Lowercase and remove Stopwords.

text="Python Is a Powerful Programming Language."
doc=nlp(text)
result=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text.lower())
print(" ".join(result))
```

```python
# 8: Display spaCy's Stopword List.

print(nlp.Defaults.stop_words)
print(len(nlp.Defaults.stop_words))
```

```python
# 9: Add a Custom Stopword.

nlp.Defaults.stop_words.add("Python")
doc=nlp("Python is a powerful programming language.")
for token in doc:
  print(token.text,"->",token.is_stop)
```

```python
# 10: Create a Stopword removal Function.

def remove_stopwords(text):
  doc=nlp(text)
  result=[]

  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.text)

  return " ".join(result)
text="The student are studying Natural Language Processing."
print(remove_stopwords(text))
```

```python
from nltk.corpus import stopwords
stop_words=set(stopwords.words("english"))
```

```python
# 1: Display NLTK stopwords.

stop_words=stopwords.words("english")
print(stop_words)
len(stop_words)
```

```python
# 2: Check whether a Word is a Stopword.

words=["the","is","Python","and","beautiful"]
for word in words:
  print(word,"->",word.lower() in stop_words)
```

```python
# 3: Extract Stopwords from a Sentence.
from nltk.tokenize import word_tokenize
nltk.download("punkt")
nltk.download("punkt_tab")

text="Python is a very sueful programming language."
tokens=word_tokenize(text)
result=[]
for word in tokens:
  if word.lower() in stop_words:
    result.append(word)
print(result)
```

```python
# 4: Remove stopwords from a sentence.

stop_words=set(stopwords.words("english"))
text="The students are learning data science."
tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(result)
```

```python
# 5: Remove stopwords and puncuation.

import string
stop_word=set(stopwords.words("english"))
text="The students are learning data science!"
tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)
print(result)
```

```python
# 6: Convert Text to Lowercase and Remove stopwords.

stop_words=set(stopwords.words("english"))
text="Python Is a Powerful Programming Language."
tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())
print(result)
```

```python
# 7: Count Stopwords in a Text.

stop_words=set(stopwords.words("english"))
text="This is a simple example to demonstrate stopword removal."
tokens=word_tokenize(text)
count=0

for word in tokens:
  if word.lower() in stop_words:
    count+=1
print("Number of stopwords:",count)
```

```python
# 8: Display stopwords by Language.

language=stopwords.fileids()
print(language)
```

```python
# 9: Add Custom stopwords in NLTK.

stop_words=set(stopwords.words("english"))
custom_words={"python","data","language"}
stop_words.update(custom_words)
text="Python is a data science language."

tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())
print(result)
```
