# Tokenization

Tokenization Using spaCy in python

Tokenization in the process of breaking text into smaller units called tokens. A token can be a word , punctuation mark number or spacial symbol.

```python
# 1. Then import sapCy
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
from IPython.core.displayhook import tokenize
# 2. Basic Tokenization

text = "Python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 3. Tokenization with token index
# Each token has an index indication its position in the document.
# token.i gives the position/index of the token.

text = "Python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.i, token.text)
```

```python
# 4. Tokenization with Token Attributes
# spaCy provides several useful attributes for each token.

text = "Python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.text, token.is_alpha, token.is_punct, token.is_space)
```

```python
# 5. Tokenization of a Sentence with punctuation
text = "Hello, how are you?"
doc = nlp(text)
for token in doc:
   print(token,text)
```

```python
# 6. Tokenization of number

text = "I bought 5 books for $10 each."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 7. Tokenization of email addresses

text = "Contact me at salmann@gmail.com"
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 8. Tokenization of URLS.

text = "Visit https://www.example.com for more information."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 9. Tokenization of Contractions

text = "I don't like Python."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 10. Tokenization of hyphnated words

text = "Data-drive decision-making is important."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 11. Removing punctuation after tokenization

text = "Python is powerful, easy, and popular!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
# 12. getting Only Alphabetic Tokens
# We can use is_alpha to select only alphabetic token.

text = "Python 3 is easy to learn!"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text)
```

```python
# 13. Converting Tokens into a List.

text = "Python is popular."
doc = nlp(text)
tokens = [token.text for token in doc]
print(tokens)
```

```python
# 14. Tokenization of multiple sentences

text = "Python is easy to learn. It is widely used in Data Science ."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 15. Tokenization and sentance segmentation
# spaCy can also indentify individual sentance.

text = "Python is easy to learn. It is widely used in Data Science."
doc = nlp(text)
for sentence in doc.sents:
  print("sentence:", sentence.text)

  for token in sentence:
    print(token.text)
  print()
```

```python
# 16. Tokenization From a Text File.

with open("sample_text_file.txt", "r",encoding="utf-8") as file:
  text = file.read()

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 17. Complete Example.

text = "Python is easy to learn! I have 5 years of experience."
doc = nlp(text)
print("Token\tAlpha\tNumber\tPunctuation")
for token in doc:
  print(token.text, "\t", token.is_alpha, "\t", token.like_num, "\t", token.is_punct)
```
