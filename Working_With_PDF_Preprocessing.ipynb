```python
# 4. Read a Specific Page
from PyPDF2 import PdfReader
reader = PdfReader("/content/Data_Science_Guide_6_Pages.pdf")
page = reader.pages[2]
print(page.extract_text())
```

```python
# 5. Search for a word in a PDF
from PyPDF2 import PdfReader
reader = PdfReader("/content/Data_Science_Guide_6_Pages.pdf")
search_word = "Data Science"

for page_number, page in enumerate(reader.pages, start=1):
  text = page.extract_text()

  if search_word.lower() in text.lower():
    print(f"'{search_word}' found on page {page_number}")
```

```python
# 6. Count words in a PDF
from PyPDF2 import PdfReader
reader = PdfReader("/content/Data_Science_Guide_6_Pages.pdf")
word_count = 0

for page in reader.pages:
  text = page.extract_text()

  if text:
    word_count += len(text.split())

print("Total words:", word_count)
```

```python
# 7. Save Extracted Text into a .txt File.
from PyPDF2 import PdfReader

reader = PdfReader("/content/Data_Science_Guide_6_Pages.pdf")

with open("pdf_text.txt", "w", encoding="utf-8") as file:
  for page_number, page in enumerate(reader.pages, start = 1):
    text = page.extract_text()

    file.write(f"\n--- Page{page_number} ---\n")
    file.write(text if text else "")

print("PDF text saved successfully.")
```

```python
# 8. Copy the First Page into Another PDF
from PyPDF2 import PdfReader, PdfWriter

# Read original PDF
reader = PdfReader("/content/Data_Science_Guide_6_Pages.pdf")

# Create new PDF
writer = PdfWriter()

# Copy first page
writer.add_page(reader.pages[0])

# Save as another PDF
with open("first_page.pdf", "wb") as output:
  writer.write(output)

print("First page copied successfully!")
```

- This method copies the actual PDF page, So it is preferable when you want to preserve the original page's formatting, images, fonts and layout.
