#**Working_With_Text_File_Advanced_Operations**

```python
# 1: creating and Writing to a Text File
# Open file in write mode

file=open("Information.txt","w")
file.write("Neha\n")
file.write("Salman\n")
file.write("Priya\n")
file.write("Amit\n")
file.write("Rahul\n")
file.write("Vikash\n")

file.close()
print("File Created Successfully.")
```

```python
# 2: Writing Multiple Lines Using writelines()

students = ["Neha\n","Salman\n","Amit\n","Priya\n"]
with open("Information.txt", "w") as file:
  file.writelines(students)
print("Data written successfully.")
```

```python
# 3: Reading the Entire File

with open("Information.txt","r") as file:
  data=file.read()
print(data)
```

```python
# 4: Reading File Line by Line

with open("Information.txt","r") as file:
  for line in file:
    print(line.strip())
```

```python
# 5: Using readline()

with open("Information.txt","r") as file:
  line1=file.readline()
  line2=file.readline()
print("First Line:",line1.strip())
print("Second Line:",line2.strip())
```

```python
# 6: Using Readlines()
with open("Information.txt","r") as file:
  lines=file.readlines()
print(lines)
```

```python
# We can also process the list Using readlines() for print context into line by line

with open("Information.txt","r") as file:
  lines=file.readlines()

for line in lines:
  print(line.strip())
```

```python
# 7: Appending data to a file

with open("Information.txt","a")as file:
  file.write("Simran\n")
  file.write("Karan\n")
print("new Students Added.")
```

```python
# 8: Writing User Input to File

name=input("Enter your name:")
age=input("Enter your age:")
with open("New_text.txt","w") as file:
  file.write("Name:"+ name +"\n")
  file.write("Age:"+ age +"\n")
print("Information Saved.")
```

```python
# 9: Counting number of Lines

with open("Information.txt","r") as file:
  lines=file.readlines()
print("number of linse:",len(lines))
```

```python
# 10. Counting Words in a Text File.split() is split character after space

with open("Information.txt","r") as file:
  data=file.read()
words=data.split()
print("Number of words:",len(words))
```

```python
# 11. Counting Characters. To include spaces

with open("Information.txt","r") as file:
  data=file.read()
print("number of Characters:",len(data))
```

```python
# To exclude spaces:

characters=data.replace(" ","").replace("\n","")
print("characters excluding space:",len(characters))
```

```python
# 12. Searching for word

search_word=input("Enter word to search:")
with open("Information.txt","r") as file:
  data=file.read()

if search_word.lower() in data.lower():
  print("Word found.")
else:
  print("Word not found.")
```

```python
# 13. Counting Occurrences of a word

search_word=input("Enter word to search:")
with open("Information.txt","r") as file:
  data=file.read()

count=data.lower().count(search_word.lower())
print("Occurrences:",count)
```

```python
# 14. Copying contents from one file to another.

with open("Information.txt","r") as source:
  data=source.read()

with open("backup.txt","w") as destination:
  destination.write(data)

print("File copied Successfully.")
```

```python
# 15. Reading only the first 10 characters

with open("Information.txt","r") as file:
  data=file.read(10)
print(data)
```
